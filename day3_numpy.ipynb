```python
import numpy as np

print("Day 3: NumPy Basics Started ")

# 1. Creating NumPy Arrays
arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([[1, 2, 3], [4, 5, 6]])

print("\n1D Array:", arr1)
print("2D Array:\n", arr2)

# 2. Array Properties
print("\nShape of arr2:", arr2.shape)
print("Data type:", arr1.dtype)

# 3. Indexing and Slicing
print("\nFirst element of arr1:", arr1[0])
print("Last 3 elements:", arr1[-3:])
print("Element at [1,2] in arr2:", arr2[1,2])
```

```python
# 4. Mathematical Operations
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print("\nAddition:", a + b)
print("Multiplication:", a * b)
print("Mean of a:", np.mean(a))
print("Sum of b:", np.sum(b))
print("Max value in a:", np.max(a))

# 5. Array-based Calculation - Student Score Example
study_hours = np.array([2, 4, 5, 6, 8])
scores = np.array([40, 60, 70, 80, 95])

print("\nAverage Study Hours:", np.mean(study_hours))
print("Average Score:", np.mean(scores))
```
