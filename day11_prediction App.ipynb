```python

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import os

print("="*50)
print(" 📚 STUDENT SCORE PREDICTION APP 📚")
print(" Powered by Codomax AI & ML Model")
print("="*50)

# 1. Load Data and Train Model
try:
    df = pd.read_csv('student_scores.csv')
except FileNotFoundError:
    print("Error: 'student_scores.csv' file not found!")
    exit()

X = df[['Study_Hours']]
y = df['Score']

model = LinearRegression()
model.fit(X, y)

def get_grade(score):
    if score >= 90: return "A+ 🔥"
    elif score >= 75: return "A 👍"
    elif score >= 60: return "B 🙂"
    elif score >= 40: return "C 😐"
    else: return "F 😢 Need to study more!"

def get_tip(hours):
    if hours < 3: return "Tip: 3 hours bahut kam hai. 5+ hours target karo."
    elif hours < 6: return "Tip: Good! 1-2 hours aur badhao to A grade pakka."
    else: return "Tip: Excellent! Bas revision karte raho."

# 2. User Input Loop
while True:
    try:
        hours = float(input("\nEnter Study Hours: "))
        if hours < 0:
            print("Hours negative nahi ho sakte!")
            continue

        # 3. Predict
        predicted_score = model.predict([[hours]])[0]
        predicted_score = max(0, min(100, predicted_score)) # 0-100 ke beech me

        grade = get_grade(predicted_score)
        tip = get_tip(hours)

        # 4. Show Result - Professional Look
        print("\n" + "="*50)
        print(f" Predicted Score: {predicted_score:.2f} / 100")
        print(f" Grade: {grade}")
        print(f" {tip}")
        print("="*50)

    except ValueError:
        print("Please sirf number daalo!")

    again = input("\nKya dobara check karna hai? (y/n): ").lower()
    if again!= 'y':
        print("\nThank you for using Student Score Predictor! Bye 👋")
        break
```
