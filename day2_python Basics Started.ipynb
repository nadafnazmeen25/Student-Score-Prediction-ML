```python
# Day 2: Python Basics
print("Day 2: Python Basics Started ")

# 1. Variables and Data Types
name = "Nazmeen"
age = 21
marks = 75.50
is_student = True

print("Name:", name, "Type:", type(name))
print("Age:", age, "Type:", type(age))
print("Marks:", marks, "Type:", type(marks))
print("Is Student:", is_student, "Type:", type(is_student))
```

```python
# 2. Operators
a = 10
b = 3
print("\nAddition:", a + b)
print("Modulus:", a % b)
print("Power:", a ** b)

# 3. Loops
print("\nFor Loop:")
for i in range(1, 6):
    print("Count:", i)

print("\nWhile Loop:")
count = 1
while count <= 3:
    print("While count:", count)
    count += 1
```

```python
# 4. Functions
def greet(student_name):
    return f"Hello {student_name}, Welcome to Day 2!"

print("\nFunction Output:", greet("Nazmeen"))

# 5. Simple Program
hours = 5
if hours >= 4:
    print("\nYou studied enough today!")
else:
    print("\nStudy more!")
```
