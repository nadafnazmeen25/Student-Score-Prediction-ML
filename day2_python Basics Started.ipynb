```python
# Day 2: Python Basics
print("Day 2: Python Basics Started ")

# 1. Variables and Data Types
name = "Nazmeen"
age = 21
marks = 75.50
is_student = True

print("Name:", name, "Type:", type(name))
print("Age:", age, "Type:", type(age))
print("Marks:", marks, "Type:", type(marks))
print("Is Student:", is_student, "Type:", type(is_student))
```

```python
# 2. Data Structures
fruits = ["apple", "banana", "mango"]  #List
marks_tuple = (25, 13, 1)              #Tuple
student = {"name": "Nazmeen", "roll no": 1325}  #Dictionary

print("\nList:", fruits)
print("Tuple:", marks_tuple)
print("Dictionary:", student)
print("access dict:", student["name"])
```

```python
# 3. NumPy Basics
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
print("\nNumPy Array:", arr)
print("Mean:", np.mean(arr))
print("Sum:", np.sum(arr))
```
